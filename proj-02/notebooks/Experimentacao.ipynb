## FWI Multiparâmetro com Inversa da Hessiana via Deep Learning

O objetivo é utilizar uma rede neural convolucional (U-Net) para aprender a aplicação da inversa da Hessiana ($H^{−1}$) sobre os gradientes de velocidade ($v$) e densidade ($\rho$), visando acelerar a convergência e reduzir o cross-talk físico entre os parâmetros no modelo **Marmousi II**.

### Estrutura de Notebooks e Conteúdo

01_Geracao_Dados_Born_Multiparametro.ipynb

Objetivo: Criar o dataset de treinamento utilizando a física do espalhamento de Born.

02_Treinamento_UNet_Hessiana.ipynb

Objetivo: Treinar a CNN para atuar como o operador de precondicionamento $H^{−1}$.

03_fwi_multiescala_comparativa.ipynb

Objetivo: Executar a inversão completa e comparar o desempenho da rede com métodos tradicionais (modelo de Barzilai-Borwein, L-BFGS)


04_analise_e_resultados.ipynb

Objetivo: Análise quantitativa (loss, SSIM) e qualitativa (primeira atualização, modelo final, sismograma, PSFs).

Análise de Cross-talk:

    Inspeção de "vazamentos" de estruturas de densidade no mapa de velocidade.
    Plotagem de Point Spreading Functions (PSFs): Inserção de uma perturbação unitária em v para ver se a rede gera resposta residual no canal de $\rho$ .

 Comparação do tempo por época e número de resoluções da equação da onda (WE solves)
